```python
import pandas as pd
from rdkit import Chem

# Function to check if phosphorus is part of a ring in the SMILES string
def has_phosphorus_in_ring(smiles):
    try:
        # Convert SMILES string to RDKit molecule object
        mol = Chem.MolFromSmiles(smiles)
        if mol is None:
            return False  # Invalid SMILES
        
        # Get all atoms in the molecule
        for atom in mol.GetAtoms():
            # Check if the atom is phosphorus
            if atom.GetSymbol() == 'P':
                # Check if phosphorus is part of a ring
                if atom.IsInRing():
                    return True
        return False
    except:
        return False

# Read the Excel file containing SMILES strings and other columns
file_path = 'descriptors.xlsx'  # Replace with your file path
df = pd.read_excel(file_path)

# Assuming the SMILES strings are in a column called 'SMILES'
if 'smiles' not in df.columns:
    raise ValueError("The column 'SMILES' was not found in the Excel file")

# Filter rows where phosphorus is part of a cyclic ring
df['P_in_ring'] = df['smiles'].apply(has_phosphorus_in_ring)

# Separate SMILES with and without phosphorus in a ring, keeping all original columns
df_with_phosphorus_in_ring = df[df['P_in_ring'] == True].copy()
df_without_phosphorus_in_ring = df[df['P_in_ring'] == False].copy()

# Save the filtered SMILES (with P in a ring) along with all columns to a new Excel file
output_file_with_ring = 'smiles_with_phosphorus_in_ring.xlsx'
df_with_phosphorus_in_ring.drop(columns=['P_in_ring'], inplace=True)  # Remove the helper column
df_with_phosphorus_in_ring.to_excel(output_file_with_ring, index=False)

# Save the filtered SMILES (without P in a ring) along with all columns to another Excel file
output_file_without_ring = 'smiles_without_phosphorus_in_ring.xlsx'
df_without_phosphorus_in_ring.drop(columns=['P_in_ring'], inplace=True)  # Remove the helper column
df_without_phosphorus_in_ring.to_excel(output_file_without_ring, index=False)

# Provide a count of how many have been removed (i.e., had phosphorus in a ring)
num_removed = len(df_with_phosphorus_in_ring)
num_remaining = len(df_without_phosphorus_in_ring)

print(f"Total SMILES processed: {len(df)}")
print(f"SMILES with phosphorus in a ring (removed): {num_removed}")
print(f"SMILES without phosphorus in a ring (remaining): {num_remaining}")
print(f"Filtered SMILES strings with phosphorus in a ring saved to {output_file_with_ring}")
print(f"SMILES without phosphorus in a ring saved to {output_file_without_ring}")
```
